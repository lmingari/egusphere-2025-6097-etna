## 2018 Etna eruption: eruptive phases

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta
```

```python
data = [
    (10.0, 11.0, 1000, 500),
    (11.0, 11.5, 3500, 8000),
    (11.5, 12.0, 6000, 20000),
    (12.0, 13.0, 4500, 10000),
    (13.0, 14.0, 1500, 750),
    (14.0, 15.0, 1200, 600),
    (15.0, 18.0, 1000, 500),
]
```

```python
df = pd.DataFrame(data, columns=["start_h", "end_h", "height_m", "emission_rate"])

base = datetime(2018, 12, 24, 0, 0)
df["start_time"] = df["start_h"].apply(lambda h: base + timedelta(hours=h))
df["end_time"] = df["end_h"].apply(lambda h: base + timedelta(hours=h))

print(df)
```

## Plot eruptive column height

```python
# build edges: start_time of each phase, plus the final end_time
times = list(df["start_time"]) + [df["end_time"].iloc[-1]]
heights = list(df["height_m"]) + [df["height_m"].iloc[-1]]

fig, ax = plt.subplots(figsize=(10, 5), dpi=200)

ax.step(times, heights, where="post", color="steelblue", linewidth=2)

ax.set_xlabel("24 Dec 2018 [UTC]")
ax.set_ylabel("Eruptive Column Height (m avl)")
ax.set_title("Eruptive Phases for the Etna experiment")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
ax.grid(True, alpha=0.3)
fig.autofmt_xdate()
```
