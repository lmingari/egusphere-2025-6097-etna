# Metrics

```python
import xarray as xr
import numpy as np
```

```python
fname = "reference.nc"
rmse_sampling = {}
for experiment in ['vae03','vae04','vae05','vae06','vae07']:
    fname_sample = f"output/sampling-{experiment}.nc"
    with xr.open_dataset(fname) as ds1, xr.open_dataset(fname_sample) as ds2:
        da1 = ds1['mean']
        da2 = ds2['samples'].mean(dim='ens')
        err = (da1-da2)**2
        mse = err.mean().item()
        rmse_sampling[experiment] = np.sqrt(mse)
```

```python
rmse_sampling
```

```python
rmse_reconstruction = {}
for experiment in ['vae03','vae04','vae05','vae06','vae07']:
    fname = f"output/recon_{experiment}.nc"
    with xr.open_dataset(fname) as ds:
        da1 = ds['reconstruction']
        da2 = ds['target']
        err = (da1-da2)**2
        mse = err.mean().item()
        rmse_reconstruction[experiment] = np.sqrt(mse)
```
