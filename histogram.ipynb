```python
import xarray as xr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fname = 'data/helens64.ens.nc'

## 1. Loading raw data and normaliation
ds = xr.open_dataset(fname)
da = ds["tephra_grn_load"]
```

```python
x = da.plot.hist(bins=40)
```

```python
# Generate x values starting from 0 up to 10
x = np.linspace(0, 400, 4000)

# Compute the natural logarithm of (x + 1e-6)
y = np.log1p(1E6*x)

# Create plot
plt.plot(x, y, label=r'$y = log1p(x)$', color='tab:blue', linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
```
